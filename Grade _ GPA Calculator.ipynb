```python
# Standard GPA mapping (Aap ise apne college/school ke hisaab se change kar sakte hain)
GRADE_SCALE = {
    'A+': 4.0, 'A': 4.0, 'A-': 3.7,
    'B+': 3.3, 'B': 3.0, 'B-': 2.7,
    'C+': 2.3, 'C': 2.0, 'C-': 1.7,
    'D+': 1.3, 'D': 1.0, 'F': 0.0
}

def calculate_gpa(courses: list) -> float:
    total_grade_points = 0.0
    total_credits = 0.0

    for grade, credits in courses:
        # Grade ke corresponding points nikalo
        points = GRADE_SCALE.get(grade)

        # Weighted points (Points * Credits)
        total_grade_points += points * credits
        total_credits += credits

    # Zero division se bachne ke liye check
    if total_credits == 0:
        return 0.0

    # Final GPA calculation: (Total Grade Points / Total Credits)
    gpa = total_grade_points / total_credits
    return gpa

def main():
    print("=== GPA / Grade Calculator ===")
    print("Valid Grades: A+, A, A-, B+, B, B-, C+, C, C-, D+, D, F")
    print("Type 'done' when you have finished adding courses.\n")

    courses = []  # Is list me hum (grade, credits) store karenge

    while True:
        # User input for grade
        grade_input = input("Enter Letter Grade (or 'done' to calculate): ").strip().upper()

        if grade_input == 'DONE':
            break

        if grade_input not in GRADE_SCALE:
            print("Error: Invalid grade entered. Please use standard letter grades (A, B, C, etc.).\n")
            continue

        # User input for credits
        try:
            credits_input = float(input(f"Enter credits for grade '{grade_input}': "))
            if credits_input <= 0:
                print("Error: Credits must be greater than 0.\n")
                continue

            # Grade aur credits ko list me add kar do
            courses.append((grade_input, credits_input))
            print(f"Added: {grade_input} ({credits_input} credits)\n")

        except ValueError:
            print("Error: Please enter a valid number for credits.\n")

    # Calculation and Final Output
    if len(courses) > 0:
        print("\n--- Final Report ---")
        final_gpa = calculate_gpa(courses)

        # Output ko 2 decimal places tak limit kiya hai
        print(f"Total Courses: {len(courses)}")
        print(f"Cumulative GPA: {final_gpa:.2f}")
    else:
        print("\nNo courses were entered.")

if __name__ == "__main__":
    main()
```
