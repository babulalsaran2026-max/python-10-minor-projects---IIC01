```python
import random

def main():
    print("=== Rock, Paper, Scissors! ===")
    print("Type 'rock', 'paper', or 'scissors' to play.")
    print("Type 'quit' to exit the game.\n")

    # State Management: Keeping track of the score across the session
    score = {
        "wins": 0,
        "losses": 0,
        "ties": 0
    }

    # Condition Mapping: What beats what? (Key beats Value)
    win_map = {
        "rock": "scissors",
        "paper": "rock",
        "scissors": "paper"
    }

    # Extract the valid choices directly from the dictionary keys
    valid_choices = list(win_map.keys())

    while True:
        user_choice = input("Your move: ").strip().lower()

        # Check for exit condition
        if user_choice == 'quit':
            break

        # Validate input
        if user_choice not in valid_choices:
            print("Invalid move! Please choose rock, paper, or scissors.\n")
            continue

        # Computer makes a random choice
        comp_choice = random.choice(valid_choices)
        print(f"Computer's move: {comp_choice}")

        # Determine the winner using our condition mapping
        if user_choice == comp_choice:
            print("Result: It's a tie! 🤝")
            score["ties"] += 1

        elif win_map[user_choice] == comp_choice:
            # If the computer's choice matches the "loser" value for the user's choice
            print("Result: You win! 🎉")
            score["wins"] += 1

        else:
            print("Result: You lose! 😢")
            score["losses"] += 1

        # Display current score
        print(f"Scoreboard -> Wins: {score['wins']} | Losses: {score['losses']} | Ties: {score['ties']}\n")

    # Final summary when the game ends
    print("\n=== Final Score ===")
    print(f"Wins:   {score['wins']}")
    print(f"Losses: {score['losses']}")
    print(f"Ties:   {score['ties']}")
    print("Thanks for playing!")

if __name__ == "__main__":
    main()
```
