```python
def calculate_checkout(base_price: float, discount_percentages: list, tax_rate: float) -> dict:
    current_price = base_price

    # Apply stacked discounts sequentially (Percentage Chain)
    for discount in discount_percentages:
        discount_amount = current_price * (discount / 100)
        current_price -= discount_amount

    subtotal = current_price
    total_saved = base_price - subtotal

    # Apply sales tax to the final subtotal
    tax_amount = subtotal * (tax_rate / 100)
    final_price = subtotal + tax_amount

    return {
        "Original Price": base_price,
        "Total Saved": total_saved,
        "Subtotal": subtotal,
        "Tax Amount": tax_amount,
        "Final Price": final_price
    }

def main():
    print("=== Discount & Sales Tax Calculator ===")
    print("Calculate final prices with stacked percentages (e.g., 20% off, then 10% off).\n")

    try:
        # 1. Get Base Price (Updated to ₹)
        base_price = float(input("Enter the original price (₹): "))
        if base_price < 0:
            print("Error: Price cannot be negative.")
            return

        # 2. Get Discounts (Allowing multiple inputs)
        discount_input = input("Enter discounts separated by spaces (e.g., 20 10 5) or press Enter for none: ").strip()

        # Convert the string of numbers into a list of floats
        if discount_input:
            discounts = [float(d) for d in discount_input.split()]
        else:
            discounts = []

        # 3. Get Sales Tax (GST)
        tax_input = input("Enter sales tax / GST % (e.g., 18 for 18%): ").strip()
        tax_rate = float(tax_input) if tax_input else 0.0

        # Calculate results
        results = calculate_checkout(base_price, discounts, tax_rate)

        # Display Receipt (Updated to ₹)
        print("\n" + "="*30)
        print(f"Original Price : ₹{results['Original Price']:,.2f}")

        if discounts:
            print(f"Total Saved    : ₹{results['Total Saved']:,.2f} (-)")
            print(f"Subtotal       : ₹{results['Subtotal']:,.2f}")

        if tax_rate > 0:
            print(f"Sales Tax/GST  : ₹{results['Tax Amount']:,.2f} (+)")

        print("-" * 30)
        print(f"FINAL PRICE    : ₹{results['Final Price']:,.2f}")
        print("="*30 + "\n")

    except ValueError:
        print("\nError: Please enter valid numerical values.")

if __name__ == "__main__":
    main()
```
