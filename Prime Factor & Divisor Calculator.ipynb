```python
import math

def get_divisors(n: int) -> list:
    """Finds all unique divisors of a number efficiently."""
    divisors = set()
    # We only need to check up to the square root of n
    limit = int(math.sqrt(n)) + 1

    for i in range(1, limit):
        if n % i == 0:
            divisors.add(i)         # i is a divisor
            divisors.add(n // i)    # The corresponding pair is also a divisor

    return sorted(list(divisors))

def get_prime_factors(n: int) -> list:
    """Finds the prime factorization of a number."""
    factors = []

    # 1. Handle the only even prime number (2)
    while n % 2 == 0:
        factors.append(2)
        n = n // 2

    # 2. Handle odd prime numbers
    # Since we factored out 2, we can skip all even numbers (step=2)
    limit = int(math.sqrt(n)) + 1
    for i in range(3, limit, 2):
        while n % i == 0:
            factors.append(i)
            n = n // i

    # 3. If n is still greater than 2, it must be a prime itself
    if n > 2:
        factors.append(n)

    return factors

def main():
    print("=== Prime Factor & Divisor Calculator ===")

    while True:
        user_input = input("\nEnter a positive integer (or 'q' to quit): ").strip()

        if user_input.lower() == 'q':
            print("Goodbye!")
            break

        try:
            num = int(user_input)
            if num <= 0:
                print("Error: Please enter a number greater than 0.")
                continue

            divisors = get_divisors(num)
            prime_factors = get_prime_factors(num)

            # A prime number has exactly 2 divisors: 1 and itself
            is_prime = (len(divisors) == 2) and (num > 1)

            print(f"\n--- Results for {num} ---")
            print(f"Is Prime?      : {'Yes ✅' if is_prime else 'No ❌'}")
            print(f"Total Divisors : {len(divisors)}")
            print(f"All Divisors   : {divisors}")

            # Format prime factorization like: 2 x 2 x 5
            if num > 1:
                factor_str = " x ".join(map(str, prime_factors))
                print(f"Prime Factors  : {factor_str}")
            else:
                print("Prime Factors  : None (1 is not a prime number)")

        except ValueError:
            print("Error: Invalid input. Please enter a whole number.")

if __name__ == "__main__":
    main()
```
