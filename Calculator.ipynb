```python
# simple calculator
choice=(input("enter your choice :"))
num1 = float(input('Enter first number: '))
num2 = float(input('Enter second number:'))
if choice=='+':
   print(num1+num2)
elif choice=='-':
   print(num1-num2)
elif choice=='*':
   print(num1*num2)
elif choice=='/':
   if num2 !=0:
     print(num1/num2)
   else:
       print("division by zero is not possible")
```
