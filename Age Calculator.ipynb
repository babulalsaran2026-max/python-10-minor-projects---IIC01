```python
from datetime import datetime

print("=== Simple Age Calculator ===")

# Get inputs from the user
name = input("Enter your name: ")
birth_year = input("Enter your birth year (e.g., 1998): ")

try:
    birth_year = int(birth_year)
    current_year = datetime.now().year

    if birth_year > current_year:
        print("\nInvalid year! Your birth year cannot be in the future.")
    else:
        age = current_year - birth_year
        print(f"\nHello {name}, you are approx. {age} years old!")

except ValueError:
    print("\nPlease enter a valid 4-digit number for the year.")
```
