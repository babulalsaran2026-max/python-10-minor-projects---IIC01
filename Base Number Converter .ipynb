```python
# base Number Converter (Binary / Hex / Decimal)
def convert_number (val_str:str, source_base:int):
#convert string representation from specified base into a standard integer
    decimal_val = int(val_str, source_base)
    return{
        'decimal':str( decimal_val),
        'binary': bin(decimal_val),    #strip '0b' prefix
        'octal':oct(decimal_val),      # strip'0o' prefix
        'hexadecimal':hex(decimal_val).upper()  # strip '0h' prefix
        # if you remove the '0b','0o',0h the result start[2:]
 }
def main():
   print("==== Base number converter ====")
   print("supported bases:2(binary),8(octal),10(decimal),16(hexadecimal)\n")

   try:
    base_input = input("Enter source base (2,8,10,16)").strip()
    source_base =int(base_input)
    if source_base not in (2,8,10,16):
      print("error:base must be (2,8,10,16)")
      return

    val_str = input (f"Enter the base-{source_base} value").strip()

    result = convert_number(val_str,source_base )
    print("\n---convert---")
    for base_name ,converted_val in result.items():
      print(f"{base_name :<12}:{converted_val}")
   except ValueError:
      print(f"\nError: '{val_str}' is not a valid base -{source_base} number.")
if __name__ == '__main__':
    main()
```
