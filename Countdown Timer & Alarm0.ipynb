```python
import time
import sys
import platform

def play_alarm():
    """Triggers an audio alert based on the operating system."""
    system = platform.system()

    if system == "Windows":
        import winsound
        # Plays a 1000Hz tone for 1000 milliseconds (1 second), twice
        winsound.Beep(1000, 1000)
        time.sleep(0.5)
        winsound.Beep(1000, 1000)

    elif system == "Darwin":
        # macOS: Uses the built-in afplay command with a default system sound
        import os
        os.system('afplay /System/Library/Sounds/Glass.aiff')
        os.system('afplay /System/Library/Sounds/Glass.aiff')

    else:
        # Linux / Other: Triggers the standard terminal bell
        print('\a\a')
        sys.stdout.flush()

def countdown_timer(total_seconds: int):
    while total_seconds > 0:
        # divmod divides the total by 60.
        # It returns the quotient (minutes) and remainder (seconds)
        mins, secs = divmod(total_seconds, 60)

        # Format variables to always show 2 digits (e.g., 05:09 instead of 5:9)
        timer_format = f"{mins:02d}:{secs:02d}"

        # \r moves the cursor back to the start of the line so it overwrites itself
        print(f"⏳ Time Remaining: {timer_format}", end="\r")

        time.sleep(1)         # Wait exactly 1 second
        total_seconds -= 1    # Reduce time by 1

    # When loop finishes, clear the line and print the final message
    print("⏰ Time's up! " + " " * 15)
    play_alarm()

def main():
    print("=== Terminal Countdown Timer ===")

    try:
        # Ask user if they are entering minutes or seconds
        time_type = input("Are you entering (m)inutes or (s)econds? ").strip().lower()

        if time_type not in ['m', 's']:
            print("Error: Please enter 'm' for minutes or 's' for seconds.")
            return

        amount = int(input("Enter the amount of time: ").strip())

        # Convert minutes to seconds if necessary
        if time_type == 'm':
            total_seconds = amount * 60
        else:
            total_seconds = amount

        print("\nStarting timer...")
        countdown_timer(total_seconds)

    except ValueError:
        print("\nError: Please enter a valid whole number for the time.")

    except KeyboardInterrupt:
        # Handles the error if you press Ctrl+C to stop the timer early
        print("\n\nTimer stopped by user.")

if __name__ == "__main__":
    main()
```
